## 1. Start with a reproducible table (8 min)

Filtering in a spreadsheet is useful, but a Python command provides a visible record of the rule used. Work from a named raw dataset and store filtered views in new variables.

```python
import pandas as pd

batches = pd.read_csv('https://raw.githubusercontent.com/vakilpriyanshu-ui/Module_5_6_Both/refs/heads/main/data/industrial_process_100.csv')
batches.head()
```

## 2. Sorting makes patterns visible (12 min)

`sort_values()` orders a table by one or more columns. Use `ascending=False` for highest first. Sorting changes the display order, not the physical order in which events happened—keep a timestamp when time sequence matters.

```python
highest_conversion = batches.sort_values('conversion_percent',ascending=False)
display(highest_conversion)

by_reactor_then_temperature = batches.sort_values(['reactor', 'temperature_c'])
display(by_reactor_then_temperature)
```

## 3. Filtering with conditions (15 min)

A filter creates a Boolean condition for every row. Place that condition inside square brackets. Use `&` for AND and `|` for OR; enclose each comparison in parentheses.

```python
r1_batches = batches[batches['operator'] == 'Bimal']
low_conversion = batches[batches['conversion_percent'] < 90]
hot_and_low = batches[(batches['temperature_c'] > 85) & (batches['conversion_percent'] < 92)]

display(r1_batches)
display(low_conversion)
display(hot_and_low)
```

**Try it (5 min):** Filter batches operated by `Bimal`. Then filter all batches with conversion at least 90%.

**Challenge:** Use `.isin(['Asha', 'Chen'])` to select batches run by either operator.

## 4. Summaries by group (12 min)

`groupby()` asks the same question for each category. Here we compare average conversion and temperature by reactor. A group average is a clue to investigate, not proof of cause and effect.

```python
reactor_summary = batches.groupby('reactor')[['temperature_c', 'conversion_percent']].mean()
display(reactor_summary)

operator_summary = batches.groupby('operator')['conversion_percent'].agg(['count', 'mean', 'min', 'max'])
display(operator_summary)
```
